```python
# SAFEZONE v7.1 FIXED — Run all ONCE (no other edits required)
# VERIFIED: Existing R2PPE source videos and COCO annotations are backed up in Google Drive.
# NOTE: A GPU is essential for this accuracy-oriented training, do not silently train on CPU.
import sys, subprocess
import torch
print('PyTorch:', torch.__version__)
print('CUDA available:', torch.cuda.is_available())
try:
    p = subprocess.run(['nvidia-smi','--query-gpu=name,memory.total','--format=csv,noheader'],
                       capture_output=True, text=True, timeout=15)
    print('NVIDIA:', (p.stdout or p.stderr).strip()[:500])
except Exception as gpu_exception:
    print('NVIDIA diagnostic:', str(gpu_exception)[:300])

if not torch.cuda.is_available():
    print('\nSAFEZONE STOPPED BEFORE COPYING 5.56 GB: NO GPU ALLOCATED TO THIS SESSION.')
    print('Colab: Runtime → Change runtime type → T4 GPU → Save → Connect.')
    print('If GPU cannot be allocated, do NOT rerun repeatedly. The training is not running.')
    print('Both video and annotation archives remain safe in MyDrive/SafeZoneAI/Datasets/R2PPE_2024.')
else:
    print('GPU READY:', torch.cuda.get_device_name(0))
    print('Starting source validation, frame extraction and YOLO training...')
    import traceback
    try:
        # SafeZone AI v7 — R2PPE Video-to-YOLO Pipeline
        # Single code-cell Colab pipeline with QA gate, safe checkpointing and video-wise split.
        # MIT-style pipeline authored for this project; underlying dataset retains its own license.

        # ====== USER CONFIGURATION ======
        MODE = 'TRAIN'                # QA reviewed: train; switch to 'AUDIT' to remake previews
        FRAME_INDEX_OFFSET = 0       # QA image reviewed; use the approved candidate offset
        TRAIN_MODEL = 'yolo11s.pt'    # accuracy-oriented; no change to best_mask_v4.pt
        IMG_SIZE = 768
        EPOCHS = 45
        BATCH = 8
        SEED = 2026
        # ================================

        from pathlib import Path
        from collections import Counter, defaultdict
        import os, re, sys, json, math, shutil, zipfile, hashlib, subprocess, random, tempfile, time
        import numpy as np
        from PIL import Image, ImageDraw, ImageFont

        assert MODE in {'AUDIT', 'TRAIN'}
        if MODE == 'TRAIN':
            assert FRAME_INDEX_OFFSET in (0, 1), 'Set FRAME_INDEX_OFFSET = 0 or 1 after reviewing the alignment sheet.'

        from google.colab import drive
        drive.mount('/content/drive', force_remount=False)

        DRIVE_BASE = Path('/content/drive/MyDrive/SafeZoneAI')
        ORIGINALS = DRIVE_BASE / 'Datasets' / 'R2PPE_2024'
        ARCHIVES = DRIVE_BASE / 'Research_Archives'
        EXP_DRIVE = DRIVE_BASE / 'Experiments' / 'R2PPE_YOLO_v7'
        EXP_DRIVE.mkdir(parents=True, exist_ok=True)
        LOCAL = Path('/content/SafeZone_R2PPE_v7')
        LOCAL.mkdir(parents=True, exist_ok=True)

        VIDEOS = LOCAL / 'R2PPE_videos.zip'
        ANNS = LOCAL / 'R2PPE_annotations.zip'
        EXPECTED_MD5 = {
            VIDEOS.name: 'b82a8bf6a5d2c00e14cc5911b4f4183e',
            ANNS.name: 'c190615219de9b9b389c705c95d0c7f0',
        }

        def digest(path, kind='md5'):
            h = hashlib.new(kind)
            with Path(path).open('rb') as f:
                for chunk in iter(lambda: f.read(16 * 1024 * 1024), b''):
                    h.update(chunk)
            return h.hexdigest()

        def verified_copy_to_drive(src, dest):
            src, dest = Path(src), Path(dest)
            dest.parent.mkdir(parents=True, exist_ok=True)
            hash_src = digest(src, 'sha256')
            if dest.exists() and dest.stat().st_size == src.stat().st_size and digest(dest, 'sha256') == hash_src:
                print('Verified in Drive:', dest.name)
                return
            temp = dest.with_name(dest.name + '.partial')
            # Recopy rather than resume unknown partially-written Drive files.
            shutil.copyfile(src, temp)
            assert temp.stat().st_size == src.stat().st_size and digest(temp, 'sha256') == hash_src
            os.replace(temp, dest)
            print('Saved and verified:', dest)

        def ensure_local(filename, destination):
            source = ORIGINALS / filename
            if destination.exists() and digest(destination) == EXPECTED_MD5[filename]:
                print('Using verified local copy:', filename)
                return
            assert source.exists(), f'Missing verified Drive backup: {source}'
            print('Copying backed-up source from Drive to Colab:', filename)
            shutil.copyfile(source, destination)
            assert digest(destination) == EXPECTED_MD5[filename], f'MD5 mismatch: {filename}'

        ensure_local('R2PPE_annotations.zip', ANNS)
        assert zipfile.is_zipfile(ANNS)
        # Video backup is already safely in Drive; only copy locally when needed.

        # ---------- Load source COCO annotations, without guessing IDs ----------
        with zipfile.ZipFile(ANNS) as za:
            jsons = []
            for member in za.namelist():
                if not member.lower().endswith('.json') or member.endswith('/'):
                    continue
                try:
                    o = json.loads(za.read(member))
                    if isinstance(o, dict) and isinstance(o.get('images'), list) and isinstance(o.get('annotations'), list):
                        jsons.append((member, o))
                except Exception as err:
                    print('JSON skipped:', member, str(err)[:100])
            assert jsons, 'No COCO JSON (images + annotations) found in R2PPE_annotations.zip.'
            # Largest single COCO manifest is expected; avoid silently combining duplicated ID namespaces.
            jsons.sort(key=lambda pair: (len(pair[1]['images']), len(pair[1]['annotations'])), reverse=True)
            member, coco = jsons[0]
            print('COCO file:', member, 'images:', len(coco['images']), 'boxes:', len(coco['annotations']))
            if len(jsons) > 1:
                print('Other COCO files:', [(m, len(o['images'])) for m, o in jsons[1:8]])
            assert len(coco['images']) >= 5000, 'Unexpectedly small COCO manifest: verify annotations before continuing.'

        cats = coco.get('categories', coco.get('classes'))
        assert isinstance(cats, list) and len(cats) >= 4, 'Unexpected COCO categories/classes schema.'
        cat_names = {}
        for idx, c in enumerate(cats):
            if isinstance(c, dict):
                cid = c.get('id', c.get('category_id', idx))
                name = c.get('name', c.get('class_name', c.get('label', '')))
            elif isinstance(c, str):
                cid, name = idx, c
            else:
                raise ValueError('Unknown category item: ' + repr(c)[:200])
            cat_names[int(cid)] = str(name)
        print('Original categories (inspect):', cat_names)

        # YOLO order follows the original SafeZone v4 three-class convention.
        CLASSES = ['with_mask', 'without_mask', 'mask_weared_incorrect']

        def map_mask_category(name):
            n = re.sub('[^a-z0-9]+', '_', str(name).lower()).strip('_')
            tokens = set(n.split('_'))
            # Official R2PPE mask abbreviations MA, MI, RC, NC.
            if n in ('ma', 'mask_absent', 'mask_ma') or ('mask' in tokens and ('absent' in tokens or 'ma' in tokens)):
                return 1
            if n in ('mi', 'mask_incomplete', 'mask_mi') or ('mask' in tokens and ('incomplete' in tokens or 'mi' in tokens)):
                return 2
            if n in ('rc', 'nc', 'mask_rc', 'mask_nc', 'regular_mask_complete', 'regular_mask_complete_rc', 'n95_complete', 'n95_complete_nc', 'n95_mask_complete', 'mask_complete'):
                return 0
            if (('mask' in tokens or 'n95' in tokens) and 'complete' in tokens and 'incomplete' not in tokens):
                return 0
            return None

        mapped = {cid: map_mask_category(name) for cid, name in cat_names.items()}
        selected = {cid: dest for cid, dest in mapped.items() if dest is not None}
        print('R2PPE -> SafeZone classes:', {cat_names[k]: CLASSES[v] for k,v in selected.items()})
        assert set(selected.values()) == {0,1,2}, 'Missing at least one desired mask class. Review source class names; abort.'
        # Important: R2PPE MA includes truly no mask AND mask dropped below mouth: cannot separate those cases automatically.

        images = {a['id']: a for a in coco['images']}
        boxes_by_image = defaultdict(list)
        mask_counts = Counter()
        for ann in coco['annotations']:
            cid = int(ann.get('category_id', -1))
            if cid not in selected or ann.get('iscrowd', 0):
                continue
            bbox = ann.get('bbox', [])
            if len(bbox) != 4 or bbox[2] <= 0 or bbox[3] <= 0:
                continue
            im = images.get(ann.get('image_id'))
            if not im:
                continue
            boxes_by_image[ann['image_id']].append((selected[cid], list(map(float, bbox))))
            mask_counts[CLASSES[selected[cid]]] += 1

        print('Mask boxes:', dict(mask_counts), 'annotated frames with masks:', len(boxes_by_image))
        assert min(mask_counts.get(cls,0) for cls in CLASSES) > 100, 'Not enough boxes of one or more mask classes.'

        # Build filename index: clinical_simulation_5_268.png -> video=clinical_simulation_5, index=268.
        by_video = defaultdict(list)
        unmatched = []
        for image_id, bbs in boxes_by_image.items():
            item = images[image_id]
            base = Path(item['file_name']).stem
            m = re.fullmatch(r'(.+)_([0-9]+)', base)
            if m is None:
                unmatched.append(item['file_name'])
                continue
            by_video[m.group(1).lower()].append({
                'image_id': str(image_id), 'file_name': item['file_name'],
                'index': int(m.group(2)), 'width': int(item['width']), 'height': int(item['height']),
                'boxes': bbs,
            })
        assert len(unmatched) < 0.03 * len(boxes_by_image), f'Unexpected video-frame names: {unmatched[:10]}'
        print('Videos represented in COCO:', len(by_video), 'samples:', {v:len(x) for v,x in sorted(by_video.items())})

        # Capture source index and taxonomy metadata before extracting any video.
        metadata = {
            'source_url': 'https://zenodo.org/records/13851664',
            'source_coco_json': member, 'num_coco_images': len(coco['images']),
            'mask_object_counts': dict(mask_counts), 'category_names': cat_names,
            'class_order': CLASSES,
            'class_semantic_caveat': 'R2PPE Mask Absent includes no-mask and mask pulled below mouth; mapped to without_mask only as violation proxy.',
            'video_frame_index_names': {v: {'min': min(i['index'] for i in items), 'max': max(i['index'] for i in items), 'annotated':len(items)} for v,items in by_video.items()},
            'dataset_license_note': 'Cite R2PPE authors and verify license for your intended application.',
        }
        (LOCAL / 'r2ppe_source_audit.json').write_text(json.dumps(metadata, indent=2, ensure_ascii=False), encoding='utf-8')
        verified_copy_to_drive(LOCAL / 'r2ppe_source_audit.json', EXP_DRIVE / 'r2ppe_source_audit.json')

        ensure_local('R2PPE_videos.zip', VIDEOS)
        assert zipfile.is_zipfile(VIDEOS)
        with zipfile.ZipFile(VIDEOS) as zv:
            members = {Path(n).stem.lower(): n for n in zv.namelist() if n.lower().endswith(('.avi','.mp4','.mov','.mkv'))}
            missing_videos = set(by_video) - set(members)
            assert not missing_videos, f'Video files do not match COCO stems: {sorted(missing_videos)[:20]}'
            print('Matched', len(by_video), 'annotated video names to saved archives.')

        import cv2

        def get_video_file(zv, stem, out_dir):
            member = members[stem]
            p = out_dir / Path(member).name
            if not p.exists():
                with zv.open(member) as f, p.open('wb') as g:
                    shutil.copyfileobj(f, g, length=16*1024*1024)
            return p

        def snapshot(cap, t):
            cap.set(cv2.CAP_PROP_POS_MSEC, max(0, t)*1000)
            ok, frame = cap.read()
            return frame if ok else None

        def draw_boxes(frame, rec):
            frame = frame.copy()
            h,w = frame.shape[:2]
            if abs(w/h - rec['width']/rec['height']) > 0.015:
                raise RuntimeError('Image/video aspect ratio mismatch! Need manual source alignment review.')
            sx = w / rec['width']; sy = h / rec['height']
            palette = [(0,155,0),(0,0,210),(0,190,220)]
            for cl,b in rec['boxes']:
                x,y,bw,bh = b
                pt1=(int(x*sx),int(y*sy))
                pt2=(int((x+bw)*sx),int((y+bh)*sy))
                cv2.rectangle(frame,pt1,pt2,palette[cl], max(2, w//700))
                cv2.putText(frame,CLASSES[cl], (pt1[0],max(15,pt1[1]-7)),cv2.FONT_HERSHEY_SIMPLEX,.6,palette[cl],2)
            return frame

        def make_audit_previews():
            rng = random.Random(SEED)
            samples = defaultdict(list)
            for v,items in sorted(by_video.items()):
                for item in items:
                    for cl,b in item['boxes']:
                        if b[2] >= 25 and b[3] >= 20:
                            samples[cl].append((v,item))
            selected_rows=[]
            for cl in (2,1,0):
                rng.shuffle(samples[cl]); chosen=[]; used=set()
                for v,item in samples[cl]:
                    if v in used:continue
                    used.add(v); chosen.append((v,item,cl))
                    if len(chosen) >= 3:break
                selected_rows.extend(chosen)
            assert len(selected_rows) >= 6, 'Not enough examples to audit alignment.'
            # Side-by-side: candidate index offset=0 versus offset=1, plus bbox overlays.
            tile_w,tile_h=480,310
            sheet=Image.new('RGB',(2*tile_w,len(selected_rows)*tile_h), 'white')
            draw=ImageDraw.Draw(sheet)
            with zipfile.ZipFile(VIDEOS) as zv:
                with tempfile.TemporaryDirectory(prefix='safezone_video_qa_') as td:
                    for row,(v,rec,cl) in enumerate(selected_rows):
                        video=get_video_file(zv,v,Path(td))
                        cap=cv2.VideoCapture(str(video))
                        if not cap.isOpened():
                            raise RuntimeError(f'Cannot decode {video.name}; stop audit.')
                        for col,offset in enumerate((0,1)):
                            # Timestamps are candidate interpretations, not verified ground truth.
                            frame=snapshot(cap,rec['index']-offset)
                            if frame is None:continue
                            frame=draw_boxes(frame, rec)
                            img=Image.fromarray(cv2.cvtColor(frame,cv2.COLOR_BGR2RGB))
                            img.thumbnail((tile_w-12,tile_h-55))
                            sheet.paste(img,(col*tile_w+(tile_w-img.width)//2,row*tile_h+4))
                            draw.text((col*tile_w+10,row*tile_h+tile_h-45),f'{v}  sec={rec["index"]-offset}  {CLASSES[cl]}',fill=(30,30,30))
                            draw.text((col*tile_w+10,row*tile_h+tile_h-22),f'candidate offset={offset}; inspect box alignment',fill=(150,60,60))
                        cap.release()
                        video.unlink(missing_ok=True)
                        print('QA preview',row+1,'/',len(selected_rows),v)
            sheet_path=LOCAL/'alignment_review_OFFSETS_0_vs_1.jpg'
            sheet.save(sheet_path,quality=90)
            verified_copy_to_drive(sheet_path, EXP_DRIVE/sheet_path.name)
            from IPython.display import display
            display(sheet)
            print('QA sheet saved:', EXP_DRIVE/sheet_path.name)
            return sheet_path

        review = EXP_DRIVE/'alignment_review_OFFSETS_0_vs_1.jpg'
        if MODE == 'AUDIT':
            print('\n=== SAFEZONE PRE-TRAINING AUDIT ===')
            print('Original R2PPE video and annotation backups verified. No training started.')
            make_audit_previews()
            print('\nNEXT: review bounding boxes and select FRAME_INDEX_OFFSET=0 or 1. Set MODE="TRAIN" only after alignment is confirmed.')
            print('R2PPE MA is an ambiguous no-mask / below-mouth label; report violation recall separately.')
        else:
            import torch
            assert torch.cuda.is_available(), 'GPU required: Runtime > Change runtime type > T4 GPU.'
            from collections import defaultdict
            assert shutil.disk_usage('/content').free > 12 * 1024**3, 'Need more than 12 GB of free Colab disk space.'
            # Video-wise split. Never random-split frames within one video.
            videos=sorted(by_video)
            video_class_counts={v:Counter(cl for item in by_video[v] for cl,_ in item['boxes']) for v in videos}
            rng=random.Random(SEED)
            candidate=None
            for attempt in range(10000):
                perm=videos.copy(); rng.shuffle(perm)
                n=len(perm)
                nv=max(3,round(n*.15)); nt=max(3,round(n*.15))
                roles={'train':perm[:n-nv-nt], 'val':perm[n-nv-nt:n-nt], 'test':perm[n-nt:]}
                sums={role: Counter() for role in roles}
                for role,vlist in roles.items():
                    for v in vlist:sums[role].update(video_class_counts[v])
                if all(all(sums[role][cl]>=20 for cl in (0,1,2)) for role in roles):
                    candidate=(roles,sums);break
            assert candidate is not None, 'Unable to construct video-disjoint 3-class split with adequate instances.'
            roles,sums=candidate
            print('Video-wise split:', roles)
            print('Target box counts:', {r:dict(c) for r,c in sums.items()})
            video_role={v:role for role,vv in roles.items() for v in vv}
            run_root=LOCAL/'dataset'/f'offset_{FRAME_INDEX_OFFSET}'
            prepared_marker=run_root/'prepared.json'
            have_complete_prepared_data = prepared_marker.exists()
            if have_complete_prepared_data:
                try:
                    prev = json.loads(prepared_marker.read_text())
                    have_complete_prepared_data = (
                        prev.get('offset') == FRAME_INDEX_OFFSET
                        and all(
                            len(list((run_root/'images'/role).glob('*.jpg'))) >= max(1, int(prev['counts'][role]))
                            and len(list((run_root/'labels'/role).glob('*.txt'))) >= max(1, int(prev['counts'][role]))
                            for role in ('train','val','test')
                        )
                    )
                except Exception:
                    have_complete_prepared_data = False
            if not have_complete_prepared_data:
                if run_root.exists():shutil.rmtree(run_root)
                for role in roles:
                    for folder in ('images','labels'):(run_root/folder/role).mkdir(parents=True,exist_ok=True)
                converted,skipped,issues=Counter(),Counter(),[]
                with zipfile.ZipFile(VIDEOS) as zv:
                    with tempfile.TemporaryDirectory(prefix='safezone_r2ppe_stream_') as td:
                        td=Path(td)
                        for vi,v in enumerate(videos):
                            video=get_video_file(zv,v,td)
                            cap=cv2.VideoCapture(str(video))
                            assert cap.isOpened(),f'Failed to decode: {video.name}'
                            fps=cap.get(cv2.CAP_PROP_FPS)
                            total=cap.get(cv2.CAP_PROP_FRAME_COUNT)
                            v_w=int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
                            v_h=int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
                            assert fps>0 and total>0, f'Missing FPS/frame-count for {video.name}'
                            sec_count=int(math.ceil(total/fps))
                            target_indices={item['index']-FRAME_INDEX_OFFSET: item for item in by_video[v]}
                            target_indices={k:item for k,item in target_indices.items() if 0<=k<sec_count}
                            frame_dir=td/'frames';frame_dir.mkdir(exist_ok=True)
                            # One frame per second. 0-based naming, lossless spatial geometry.
                            cmd=['ffmpeg','-hide_banner','-loglevel','error','-y','-i',str(video),
                                 '-vf','fps=1','-start_number','0','-q:v','2',str(frame_dir/'f_%06d.jpg')]
                            subprocess.run(cmd,check=True)
                            actual_files=list(frame_dir.glob('f_*.jpg'))
                            if not actual_files:
                                raise RuntimeError(f'No frames decoded from video {v}.')
                            if len(actual_files) < max(1, int(sec_count * 0.90)):
                                print(f'NOTICE: decoded frame count for {v}: {len(actual_files)}; metadata estimate {sec_count}. Verifying actual annotated indices individually.')
                            for second,rec in sorted(target_indices.items()):
                                src=frame_dir/f'f_{second:06d}.jpg'
                                if not src.exists():
                                    skipped['missing_annotated_frame']+=1;continue
                                destname=f'{v}_{rec["index"]:06d}'
                                role=video_role[v]
                                im=cv2.imread(str(src))
                                if im is None:
                                    skipped['bad_frame']+=1;continue
                                h,w=im.shape[:2]
                                iw,ih=rec['width'],rec['height']
                                if abs((w/h)-(iw/ih))>0.015:
                                    issues.append({'file':rec['file_name'],'why':'video/image aspect ratio mismatch'})
                                    skipped['aspect_mismatch']+=1;continue
                                labels=[]
                                for cl,(x,y,bw,bh) in rec['boxes']:
                                    x1=max(0,min(iw,x));y1=max(0,min(ih,y))
                                    x2=max(0,min(iw,x+bw));y2=max(0,min(ih,y+bh))
                                    bw2,bh2=x2-x1,y2-y1
                                    if bw2<3 or bh2<3:continue
                                    xc=(x1+x2)/(2*iw); yc=(y1+y2)/(2*ih)
                                    rw=bw2/iw;rh=bh2/ih
                                    labels.append(f'{cl} {xc:.7f} {yc:.7f} {rw:.7f} {rh:.7f}')
                                if not labels:
                                    skipped['no_valid_mask_label']+=1;continue
                                dst=run_root/'images'/role/(destname+'.jpg')
                                shutil.copyfile(src,dst)
                                (run_root/'labels'/role/(destname+'.txt')).write_text('\n'.join(labels)+'\n')
                                converted[role]+=1
                            cap.release()
                            shutil.rmtree(frame_dir,ignore_errors=True)
                            video.unlink(missing_ok=True)
                            print(f'[{vi+1}/{len(videos)}] {v}: {converted[video_role[v]]} {video_role[v]} images cumulative')
                assert not issues, f'{len(issues)} frame dimensions mismatched. Abort: {issues[:5]}'
                assert all(converted[r]>=40 for r in roles), f'Insufficient frames in split: {converted}'
                assert skipped['missing_annotated_frame'] < 0.05 * sum(map(len,by_video.values())), 'Too many missing frames. Check offset.'
                marker={'roles':roles,'counts':dict(converted),'skipped':dict(skipped),
                        'offset':FRAME_INDEX_OFFSET,'source_annotation_md5':EXPECTED_MD5[ANNS.name],
                        'source_video_md5':EXPECTED_MD5[VIDEOS.name],'class_mapping':{cat_names[k]:CLASSES[x] for k,x in selected.items()},
                        'warning':'Sampled frames from AVI at 1fps. Visual alignment must be verified before relying on results.',
                        'semantic_warning':'Mask Absent includes down-below-mouth masks in R2PPE'}
                prepared_marker.write_text(json.dumps(marker,indent=2),encoding='utf-8')
            else:
                marker=json.loads(prepared_marker.read_text())
                assert marker['offset']==FRAME_INDEX_OFFSET
                print('Reusing prepared video-disjoint images:',marker['counts'])
            data_yaml=run_root/'data.yaml'
            data_yaml.write_text('path: '+str(run_root)+'\ntrain: images/train\nval: images/val\ntest: images/test\nnames:\n  0: with_mask\n  1: without_mask\n  2: mask_weared_incorrect\n',encoding='utf-8')
            verified_copy_to_drive(prepared_marker,EXP_DRIVE/f'prepared_offset_{FRAME_INDEX_OFFSET}.json')
            verified_copy_to_drive(data_yaml,EXP_DRIVE/f'data_offset_{FRAME_INDEX_OFFSET}.yaml')
            # Lightweight preview of prepared frames and their labels.
            sample_p=LOCAL/'prepared_yolo_sample.jpg'
            if not sample_p.exists():
                chosen=[]
                for role in ('train','val','test'):
                    pics=sorted((run_root/'images'/role).glob('*.jpg'))
                    random.Random(SEED).shuffle(pics)
                    chosen.extend((role,p) for p in pics[:6])
                sheet=Image.new('RGB',(3*440,6*300),'white');d=ImageDraw.Draw(sheet)
                for i,(role,p) in enumerate(chosen):
                    im=Image.open(p).convert('RGB');w,h=im.size
                    cv2im=np.asarray(im).copy()
                    label=run_root/'labels'/role/(p.stem+'.txt')
                    for line in label.read_text().splitlines():
                        cl,xc,yc,bw,bh=map(float,line.split())
                        x1,y1=int((xc-bw/2)*w),int((yc-bh/2)*h)
                        x2,y2=int((xc+bw/2)*w),int((yc+bh/2)*h)
                        cv2.rectangle(cv2im,(x1,y1),(x2,y2),[(0,155,0),(200,0,0),(0,150,200)][int(cl)],max(2,w//700))
                    img=Image.fromarray(cv2im);img.thumbnail((430,270))
                    col,row=i//6,i%6
                    sheet.paste(img,(col*440+(440-img.width)//2,row*300))
                    d.text((col*440+8,row*300+275),role+' '+p.name[:24],fill='black')
                sheet.save(sample_p,quality=88)
            verified_copy_to_drive(sample_p,EXP_DRIVE/sample_p.name)

            subprocess.check_call([sys.executable,'-m','pip','-q','install','ultralytics>=8.3,<9'])
            from ultralytics import YOLO
            project=LOCAL/'runs'
            run_name=f'r2ppe_yolo11s_{IMG_SIZE}_offset{FRAME_INDEX_OFFSET}'
            run_dir=project/run_name
            ckpt_dir=EXP_DRIVE/'checkpoints';ckpt_dir.mkdir(exist_ok=True)
            model=YOLO(TRAIN_MODEL)
            print('CUDA verified:', torch.cuda.get_device_name(0), '| preparing YOLO11s training')
            def sync_epoch(trainer):
                # Save resumable training state in Drive every three epochs.
                if (int(trainer.epoch)+1)%3 == 0:
                    last=Path(trainer.save_dir)/'weights'/'last.pt'
                    if last.exists():
                        try:
                            verified_copy_to_drive(last,ckpt_dir/'last.pt')
                        except Exception as backup_error:
                            print('WARNING: checkpoint backup failed, local training continues:', backup_error)
            model.add_callback('on_fit_epoch_end',sync_epoch)
            print('Starting YOLO experiment. v4 checkpoint is never overwritten.')
            model.train(data=str(data_yaml),epochs=EPOCHS,imgsz=IMG_SIZE,batch=BATCH,
                        device=0,workers=2,patience=12,seed=SEED,deterministic=True,
                        optimizer='AdamW',cos_lr=True,project=str(project),name=run_name,
                        exist_ok=True,plots=True,save=True)
            best=run_dir/'weights'/'best.pt';last=run_dir/'weights'/'last.pt'
            assert best.exists(), 'No best YOLO weights; training incomplete.'
            for fn in (best,last):
                if fn.exists(): verified_copy_to_drive(fn,EXP_DRIVE/f'{run_name}_{fn.name}')
            model=YOLO(str(best))
            # Test is evaluated only here, after training and val-based checkpoint selection.
            val_metrics=model.val(data=str(data_yaml),split='val',imgsz=IMG_SIZE,device=0,plots=True)
            test_metrics=model.val(data=str(data_yaml),split='test',imgsz=IMG_SIZE,device=0,plots=True)
            def scores_by_class(m):
                result = {k: None for k in CLASSES}
                for class_idx, ap in zip(m.box.ap_class_index, m.box.ap50):
                    result[CLASSES[int(class_idx)]] = float(ap)
                return result

            summary={
                'model':TRAIN_MODEL,'imgsz':IMG_SIZE,'best_checkpoint':str(EXP_DRIVE/f'{run_name}_best.pt'),
                'split_by':'source video', 'source_videos':roles,
                'train_val_test_images':marker['counts'],
                'val_map50':float(val_metrics.box.map50), 'val_map50_95':float(val_metrics.box.map),
                'test_map50':float(test_metrics.box.map50), 'test_map50_95':float(test_metrics.box.map),
                'val_ap50_per_class':scores_by_class(val_metrics),
                'test_ap50_per_class':scores_by_class(test_metrics),
                'caveats':['Never compare classifier accuracy directly to detection mAP.',
                           'Train/val/test split by video, not assured person-disjoint; all videos from same simulation site.',
                           'Mask Absent includes mask worn below the mouth: not strictly absent.',
                           'Requires independent factory gate video validation before deployment.',
                           'Video frame/annotation alignment chosen by manual visual QA.'],
            }
            report=LOCAL/'r2ppe_yolo_v7_results.json'
            report.write_text(json.dumps(summary,indent=2,ensure_ascii=False),encoding='utf-8')
            verified_copy_to_drive(report,EXP_DRIVE/report.name)
            for fn in ('results.csv','confusion_matrix.png','confusion_matrix_normalized.png'):
                p=run_dir/fn
                if p.exists():verified_copy_to_drive(p,EXP_DRIVE/f'{run_name}_{fn}')
            print('\nSAFEZONE YOLO V7 RESEARCH COMPLETE')
            print(json.dumps(summary,indent=2,ensure_ascii=False))
            print('Retain best_mask_v4.pt as production baseline until real-video evaluation.')
    except Exception as error:
        report = 'SAFEZONE V7.1 FAILURE\n' + traceback.format_exc()
        log_path = Path('/content/SafeZone_v7_1_last_error.txt')
        log_path.write_text(report, encoding='utf-8')
        print('\nSAFEZONE ERROR REPORT: ' + str(log_path))
        print(report[-6000:])
        drive_path = Path('/content/drive/MyDrive/SafeZoneAI/Experiments/R2PPE_YOLO_v7')
        if drive_path.is_dir():
            try:
                import shutil
                shutil.copy2(log_path, drive_path / log_path.name)
                print('Error report preserved in Drive.')
            except Exception as save_error:
                print('Could not copy error report to Drive:', repr(save_error))
        raise
```
