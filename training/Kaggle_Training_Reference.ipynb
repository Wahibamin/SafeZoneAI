## 3 — الاتصال الآمن بـ Google Drive (بدون تسجيل الدخول كل مرة)

**لماذا rclone؟** لأن Kaggle لا يوفر `drive.mount()` الخاص بـ Google Colab. سنستعمل OAuth المأخوذ على جهازك ثم نحفظ إعداد الاتصال بصيغة Base64 **داخل Kaggle Secrets**. Base64 ليس تشفيرًا أمنيًا؛ الحماية تأتي من كون السر مخفيًا في Kaggle Secrets.

### إعداد المفتاح لأول مرة على Windows

1. ثبّت `rclone` على حاسوبك: https://rclone.org/downloads/  (يمكن `winget install Rclone.Rclone`).
2. افتح PowerShell ثم شغّل `rclone config`، اختر `n` لإنشاء اتصال باسم **`gdrive`**، وحدد **Google Drive**، واختر `drive` للصلاحية، ووافق على تسجيل الدخول بحساب Google في المتصفح، ثم احفظ.
3. اعثر على ملف الإعدادات باستخدام `rclone config file`؛ عادة يوجد في `%APPDATA%\rclone\rclone.conf`.
4. في PowerShell نفّذ:

```powershell
$cfg = "$env:APPDATA\rclone\rclone.conf"
[Convert]::ToBase64String([IO.File]::ReadAllBytes($cfg)) | Set-Clipboard
```

5. افتح Kaggle → **Add-ons → Secrets**، أنشئ `RCLONE_CONFIG_B64` والصق ما في الحافظة وفعّل صلاحية الدفتر.

**تنبيه:** لا تشارك محتوى السر أو ملف `rclone.conf` مع أحد. لا تضفه إلى Google Drive كملف نصي عام. إذا سجّلت دخولك بحساب آخر فأعد توليد السر.

````python
# تثبيت rclone الرسمي محليًا من ZIP مع التحقق وإعادة المحاولة عند انقطاع الإنترنت
RCLONE = shutil.which('rclone')
if not RCLONE:
    machine = platform.machine().lower()
    arch = 'amd64' if machine in ('x86_64', 'amd64') else 'arm64' if machine in ('aarch64','arm64') else None
    if not arch:
        raise RuntimeError(f'بنية جهاز غير مدعومة لتثبيت rclone: {machine}')
    link = f'https://downloads.rclone.org/rclone-current-linux-{arch}.zip'
    target = Path('/tmp/safezone-rclone')
    target.mkdir(exist_ok=True)
    archive = target / 'rclone.zip'
    valid_archive = False
    if archive.exists():
        try:
            with zipfile.ZipFile(archive) as z:
                valid_archive = (z.testzip() is None)
        except Exception:
            valid_archive = False
    if not valid_archive:
        archive.unlink(missing_ok=True)
        partial = target / 'rclone.zip.part'
        for attempt in range(1, 4):
            partial.unlink(missing_ok=True)
            try:
                print(f'تنزيل rclone الرسمي (محاولة {attempt}/3)...')
                urllib.request.urlretrieve(link, partial)
                with zipfile.ZipFile(partial) as z:
                    bad = z.testzip()
                    if bad:
                        raise RuntimeError(f'أرشيف rclone تالف: {bad}')
                partial.replace(archive)
                break
            except Exception:
                partial.unlink(missing_ok=True)
                if attempt == 3:
                    raise
                time.sleep(5 * attempt)
    with zipfile.ZipFile(archive) as z:
        members = [m for m in z.namelist() if m.endswith('/rclone')]
        if len(members) != 1:
            raise RuntimeError('تعذر تحديد rclone من الأرشيف الرسمي')
        executable = target / 'rclone'
        executable.write_bytes(z.read(members[0]))
        executable.chmod(0o755)
    RCLONE = str(executable)
print(subprocess.run([RCLONE, 'version'], text=True, capture_output=True, check=True).stdout.splitlines()[0])
````

````python
# استخراج ملف الربط من Secrets إلى /tmp فقط، وليس إلى /kaggle/working
try:
    from kaggle_secrets import UserSecretsClient
    b64_config = UserSecretsClient().get_secret('RCLONE_CONFIG_B64')
except Exception as exc:
    raise RuntimeError('لم يتم العثور على سر RCLONE_CONFIG_B64 أو لم تسمح لهذا الدفتر باستعماله. راجع خلية الإعداد.') from exc

try:
    config_bytes = base64.b64decode(''.join(b64_config.split()), validate=True)
    decoded = config_bytes.decode('utf-8-sig')
    if f'[{RCLONE_REMOTE}]' not in decoded or 'type = drive' not in decoded:
        raise ValueError('اتصال Google Drive المطلوب غير موجود في الإعداد')
except Exception as exc:
    raise RuntimeError('قيمة RCLONE_CONFIG_B64 غير صالحة. أعد توليدها من rclone.conf.') from exc
finally:
    del b64_config

SECRET_DIR = Path('/tmp/safezone-private')
SECRET_DIR.mkdir(mode=0o700, parents=True, exist_ok=True)
os.chmod(SECRET_DIR, 0o700)
RCLONE_CONFIG = SECRET_DIR / 'rclone.conf'
RCLONE_CONFIG.write_bytes(config_bytes)
os.chmod(RCLONE_CONFIG, 0o600)
del config_bytes, decoded
print('✅ ربط rclone جاهز (الأسرار غير معروضة).')
````

## 4 — أدوات الرفع والاسترجاع والتحقق

لا نحذف شيئًا من Drive خلال الرفع. تُرفع نقطة Epoch باسم مستقل، ثم نحدّث ملف `latest.json` **بعد** نجاح النقل. أثناء الاستعادة نتحقق من **SHA-256** وحجم ملف النقطة. سيتم الاحتفاظ بعدة نقاط سحابية لتجنب فقدان آخر حالة سليمة.

````python
def run_rclone(args, timeout=SYNC_TIMEOUT_SECONDS, check=True):
    cmd = [RCLONE, '--config', str(RCLONE_CONFIG),
           '--retries', '2', '--low-level-retries', '2',
           '--contimeout', '15s', '--timeout', '45s',
           '--transfers', '1', '--checkers', '2'] + [str(a) for a in args]
    out = subprocess.run(cmd, text=True, capture_output=True, timeout=timeout)
    if check and out.returncode:
        # لا تطبع معلومات حساسة من rclone.conf؛ رسائل الخطأ فقط
        raise RuntimeError('rclone failed: ' + (out.stderr or out.stdout)[-500:])
    return out

def cloud_paths():
    out = run_rclone(['lsf', CLOUD_RUN, '-R', '--files-only', '--format', 'p'])
    return {line.strip() for line in out.stdout.splitlines() if line.strip()}

def sha256_file(path, block=8 * 1024 * 1024):
    h = hashlib.sha256()
    with open(path, 'rb') as stream:
        while True:
            data = stream.read(block)
            if not data: break
            h.update(data)
    return h.hexdigest()

def write_json(path, payload):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    tmp = path.with_suffix(path.suffix + '.writing')
    tmp.write_text(json.dumps(payload, ensure_ascii=False, indent=2, sort_keys=True), encoding='utf-8')
    tmp.replace(path)

def upload_file(local_path, remote_rel):
    local_path = Path(local_path)
    if not local_path.is_file(): raise FileNotFoundError(local_path)
    return run_rclone(['copyto', str(local_path), f'{CLOUD_RUN}/{remote_rel}'])

def upload_atomic(local_path, remote_rel):
    """Upload to a temporary remote name, then publish the completed file."""
    temp_rel = remote_rel + '.uploading'
    upload_file(local_path, temp_rel)
    # Remove an older destination, then publish the completed upload.
    run_rclone(['deletefile', f'{CLOUD_RUN}/{remote_rel}'], check=False)
    run_rclone(['moveto', f'{CLOUD_RUN}/{temp_rel}', f'{CLOUD_RUN}/{remote_rel}'])

def download_file(remote_rel, local_path):
    local_path = Path(local_path)
    local_path.parent.mkdir(parents=True, exist_ok=True)
    # تحميل مؤقت، لا نستبدل النسخة الحالية حتى ينجح النقل
    tmp = local_path.with_suffix(local_path.suffix + '.downloading')
    try:
        run_rclone(['copyto', f'{CLOUD_RUN}/{remote_rel}', str(tmp)])
        if not tmp.exists() or not tmp.stat().st_size: raise IOError('ملف فارغ بعد التنزيل')
        tmp.replace(local_path)
    finally:
        tmp.unlink(missing_ok=True)
    return local_path

def log_event(event, **kwargs):
    p = LOCAL_REPORT_DIR / 'events.jsonl'
    with open(p, 'a', encoding='utf8') as f:
        f.write(json.dumps({'at': datetime.datetime.now(datetime.timezone.utc).isoformat(), 'event':event, **kwargs}, ensure_ascii=False) + '\n')

# فحص اتصال ذهاب وإياب قبل أي تدريب
smoke_local = SECRET_DIR / 'drive_test.txt'
smoke_local.write_text('SafeZoneAI: ' + RUN_ID, encoding='utf8')
try:
    upload_file(smoke_local, 'meta/_connection_test.txt')
    smoke_out = download_file('meta/_connection_test.txt', SECRET_DIR / 'drive_test_return.txt')
    assert smoke_out.read_bytes() == smoke_local.read_bytes()
except Exception as exc:
    if not ALLOW_TRAINING_WITHOUT_DRIVE:
        raise RuntimeError('❌ Google Drive غير جاهز للكتابة والاسترجاع؛ التدريب متوقف حفاظًا على النتائج.') from exc
print('✅ تجربة الكتابة والقراءة من Google Drive نجحت.')
````

## 5 — محاولة استرجاع آخر Epoch من Google Drive

تنفّذ هذه الخلية **قبل التدريب** في كل جلسة. إذا كان `latest.json` ناقصًا بسبب انقطاع أثناء الحفظ، يفحص الدفتر الملفات المؤرشفة من الأحدث للأقدم. لا يسمح باستئناف ملف حجمه أو بصمته غير صحيحة.

````python
def remote_checkpoint_records(paths):
    epochs = []
    for rel in paths:
        m = re.fullmatch(r'checkpoints/epoch_(\d+)\.json', rel)
        if m:
            ep = int(m.group(1))
            pt = f'checkpoints/epoch_{ep:04d}.pt'
            if pt in paths: epochs.append((ep, rel, pt))
    return sorted(epochs, reverse=True)

def restore_cloud_checkpoint():
    files = cloud_paths()
    if 'meta/run_config.json' in files:
        download_file('meta/run_config.json', CONFIG_DIR / f'{RUN_ID}_remote_config.json')
    records = remote_checkpoint_records(files)
    if not records:
        # إذا وُجدت أوزان سحابية أو مؤشر تقدم فلا نفترض أن التجربة جديدة
        partial = [x for x in files if x.startswith('checkpoints/') or x == 'meta/latest.json' or x.startswith('weights/')]
        if partial:
            raise RuntimeError('⚠️ توجد آثار تدريب سحابي لكن لا توجد نقطة Epoch كاملة وموثقة. لن نبدأ من الصفر: راجع الملفات أو استعدها يدويًا.')
        print('ℹ️ لا توجد نقاط تدريب سحابية قابلة للاستئناف لهذه التجربة.')
        return None

    for ep, meta_name, weight_name in records:
        try:
            remote_meta_path = SECRET_DIR / f'checkpoint_epoch_{ep}.json'
            download_file(meta_name, remote_meta_path)
            meta = json.loads(remote_meta_path.read_text(encoding='utf8'))
            if meta['epoch'] != ep or meta['run_id'] != RUN_ID: raise ValueError('بيانات checkpoint لا تطابق التجربة')
            if meta['file'] != weight_name: raise ValueError('موقع checkpoint غير متطابق')
            candidate = SECRET_DIR / f'epoch_{ep}.pt'
            download_file(weight_name, candidate)
            if candidate.stat().st_size != meta['bytes'] or sha256_file(candidate) != meta['sha256']:
                raise ValueError('فشل فحص SHA256 أو الحجم')
            WEIGHTS_DIR.mkdir(parents=True, exist_ok=True)
            restored = WEIGHTS_DIR / 'last.pt'
            preserve_local = False
            if restored.is_file():
                try:
                    local_ck = torch.load(restored, map_location='cpu', weights_only=False)
                    local_epoch = int(local_ck.get('epoch', -1)) + 1
                    preserve_local = bool(local_ck.get('optimizer') is not None and local_epoch > ep)
                    del local_ck
                except Exception:
                    pass
            if preserve_local:
                print(f'✅ وجدت محليًا Epoch {local_epoch} وهو أحدث من Drive ({ep})؛ لن أستبدله.')
            else:
                shutil.copy2(candidate, restored)
            candidate.unlink(missing_ok=True)
            if not preserve_local:
                for fname in ['args.yaml','results.csv']:
                    cloud_name = f'run/{fname}'
                    if cloud_name in files:
                        download_file(cloud_name, RUN_DIR / fname)
                if 'weights/best.pt' in files:
                    try:
                        download_file('weights/best.pt', WEIGHTS_DIR / 'best.pt')
                    except Exception as ex:
                        print('⚠️ تعذر تنزيل best.pt، لكن last.pt سليم:', type(ex).__name__)
            else:
                print('ℹ️ احتفظنا أيضًا بملفات النتائج والأوزان المحلية الأحدث.')
            print(f'✅ استعادة Epoch {ep} من Drive والتحقق من SHA-256.')
            log_event('restored', epoch=ep)
            return meta
        except Exception as exc:
            print(f'⚠️ نقطة Epoch {ep} غير صالحة أو تعذر تنزيلها؛ محاولة الأقدم: {type(exc).__name__}')
    raise RuntimeError('توجد نقاط محفوظة على Drive، ولكن لم تنجح استعادة أي نقطة سليمة. لن نعيد التدريب تلقائيًا.')

CLOUD_RESTORED_META = restore_cloud_checkpoint()
````

## 6 — تحميل مجموعة البيانات دون المساس بالـ Checkpoints

**إذا كانت البيانات ضمن Kaggle Input** فلا تحتاج لإعادة التنزيل في الجلسات الجديدة. **إذا كانت على Roboflow** ستُنزّل إلى مجلد ثابت في كل جلسة جديدة (إلا إذا كانت موجودة بالفعل). التخزين السحابي هنا مخصص للتدريب؛ يمكنك وضع نسخة ZIP من البيانات على Drive واختيار `drive_zip` لتفادي Roboflow لاحقًا.

````python
def find_dataset_yaml(folder):
    folder = Path(folder)
    candidates = [p for p in folder.rglob('data.yaml') if p.is_file()]
    if not candidates:
        raise FileNotFoundError(f'لم أجد data.yaml في {folder}')
    return sorted(candidates, key=lambda p: (not (p.parent/'train'/'images').is_dir(), len(p.parts)))[0]

if DATA_SOURCE == 'roboflow':
    desired = DATA_DIR / f'{ROBOFLOW_PROJECT}_v{ROBOFLOW_VERSION}'
    existing_yamls = list(desired.rglob('data.yaml'))
    existing_layout_ok = any(
        (yaml_path.parent / 'train' / 'images').is_dir()
        and ((yaml_path.parent / 'valid' / 'images').is_dir() or (yaml_path.parent / 'val' / 'images').is_dir())
        for yaml_path in existing_yamls
    )
    if not existing_layout_ok:
        try:
            api_key = UserSecretsClient().get_secret('ROBOFLOW_API_KEY')
        except Exception as exc:
            raise RuntimeError('يرجى إنشاء سر ROBOFLOW_API_KEY في Kaggle وإتاحة وصول الدفتر إليه، أو اختر kaggle_input.') from exc
        from roboflow import Roboflow
        print('تنزيل بيانات Roboflow؛ إذا كان هناك تنزيل سابق غير مكتمل فسيُعاد تنزيله بأمان...')
        dataset = (Roboflow(api_key=api_key)
                   .workspace(ROBOFLOW_WORKSPACE)
                   .project(ROBOFLOW_PROJECT)
                   .version(ROBOFLOW_VERSION)
                   .download('yolov8', location=str(desired), overwrite=True))
        del api_key
    SOURCE_YAML = find_dataset_yaml(desired)
elif DATA_SOURCE == 'kaggle_input':
    if KAGGLE_DATA_YAML.strip():
        SOURCE_YAML = Path(KAGGLE_DATA_YAML.strip())
    else:
        found = list(Path('/kaggle/input').rglob('data.yaml'))
        if len(found) != 1:
            raise RuntimeError(f'وجدت {len(found)} ملفات data.yaml في Kaggle Input؛ حدد KAGGLE_DATA_YAML بشكل صريح في خلية الإعدادات.')
        SOURCE_YAML = found[0]
    if not SOURCE_YAML.is_file():
        raise FileNotFoundError(SOURCE_YAML)
else:
    desired = DATA_DIR / 'drive_dataset'
    desired.mkdir(parents=True, exist_ok=True)
    downloaded_zip = desired / 'dataset.zip'
    zip_ok = False
    if downloaded_zip.is_file():
        try:
            with zipfile.ZipFile(downloaded_zip) as archive:
                zip_ok = (archive.testzip() is None)
        except Exception:
            zip_ok = False
    if not zip_ok:
        downloaded_zip.unlink(missing_ok=True)
        # مصدر Drive DATA_ZIP يكون نسبة إلى remote لا إلى CLOUD_RUN
        run_rclone(['copyto', f'{RCLONE_REMOTE}:{DRIVE_DATA_ZIP.lstrip("/")}', str(downloaded_zip)], timeout=3600)
        with zipfile.ZipFile(downloaded_zip) as archive:
            bad = archive.testzip()
            if bad:
                downloaded_zip.unlink(missing_ok=True)
                raise RuntimeError(f'ملف ZIP الذي نُزّل من Drive تالف عند: {bad}')
    unpack_dir = desired / 'unpacked'
    marker = unpack_dir / '.extract_complete'
    if not marker.is_file() or not list(unpack_dir.rglob('data.yaml')):
        shutil.rmtree(unpack_dir, ignore_errors=True)
        unpack_dir.mkdir(parents=True, exist_ok=True)
        with zipfile.ZipFile(downloaded_zip) as archive:
            for item in archive.infolist():
                resolved = (unpack_dir / item.filename).resolve()
                if not resolved.is_relative_to(unpack_dir.resolve()):
                    raise RuntimeError('ZIP يحتوي مسارًا غير آمن')
            archive.extractall(unpack_dir)
        if not list(unpack_dir.rglob('data.yaml')):
            raise FileNotFoundError('تم فك ZIP لكن لم نعثر على data.yaml')
        marker.write_text('ok', encoding='utf8')
    SOURCE_YAML = find_dataset_yaml(unpack_dir)

DATASET_ROOT = SOURCE_YAML.parent.resolve()
print('✅ بيانات:', SOURCE_YAML)
print('مجلد المصدر:', DATASET_ROOT)
````

## 7 — تدقيق الفئات والملصقات، وإنشاء `data.yaml` ثابت

**مهم علميًا:** كشف `helmet` لا يعني القدرة على كشف `no_helmet`؛ نراجع قائمة الفئات قبل المقارنة مع متطلبات SafeZoneAI. الدفتر يرفض تغيير ترتيب الفئات أو تقسيم البيانات في منتصف التجربة.

````python
raw = yaml.safe_load(SOURCE_YAML.read_text(encoding='utf-8'))
raw_names = raw.get('names')
if isinstance(raw_names, dict):
    ids = [int(k) for k in raw_names]
    if sorted(ids) != list(range(len(ids))): raise ValueError('معرّفات فئات غير متسلسلة في data.yaml')
    NAMES = [str(raw_names[i] if i in raw_names else raw_names[str(i)]) for i in range(len(ids))]
elif isinstance(raw_names, list):
    NAMES = [str(n) for n in raw_names]
else:
    raise ValueError('data.yaml يجب أن يحتوي names كقائمة أو قاموس')
if not NAMES or len(set(NAMES)) != len(NAMES): raise ValueError('فئات مكررة أو فارغة')

# استخدام المجلدات الفعلية، لأن بعض ملفات Roboflow تكتب ../train/images بشكل غير قابل للنقل
split_dirs = {}
for split, options in {'train':['train'], 'val':['valid','val'], 'test':['test']}.items():
    for dirname in options:
        pp = DATASET_ROOT / dirname / 'images'
        if pp.is_dir():
            split_dirs[split] = pp
            break
    if split in ('train','val') and split not in split_dirs:
        raise RuntimeError(f'لم أجد مجلد صور {split} تحت {DATASET_ROOT}')

TRAIN_YAML = CONFIG_DIR / f'{RUN_ID}_data.yaml'
config = {'path': str(DATASET_ROOT), 'train': str(split_dirs['train']),
          'val': str(split_dirs['val']), 'nc': len(NAMES), 'names': NAMES}
if 'test' in split_dirs: config['test'] = str(split_dirs['test'])
TRAIN_YAML.write_text(yaml.safe_dump(config, sort_keys=False, allow_unicode=True), encoding='utf-8')

EXTS = {'.jpg','.jpeg','.png','.bmp','.webp','.tif','.tiff'}
IMAGE_FILES = {split: sorted([p for p in folder.rglob('*') if p.suffix.lower() in EXTS]) for split, folder in split_dirs.items()}
for split in ('train','val'):
    if not IMAGE_FILES[split]: raise RuntimeError(f'لا توجد صور في {split_dirs[split]}')

for split, files in IMAGE_FILES.items():
    labels_dir = split_dirs[split].parent / 'labels'
    if split in ('train', 'val') and not labels_dir.is_dir():
        raise RuntimeError(f'لا يوجد مجلد labels مطلوب للتدريب/التحقق: {labels_dir}')
    if not labels_dir.is_dir():
        print(f'{split}: صور {len(files):,} | لا توجد تسميات (مسموح لمجموعة test غير المستخدمة أثناء التدريب)')
        continue
    absent = 0
    for img in files:
        label = labels_dir / img.relative_to(split_dirs[split]).with_suffix('.txt')
        if not label.exists(): absent += 1
    print(f'{split}: صور {len(files):,} | ملصقات ناقصة {absent:,}')
    if split in ('train', 'val') and absent == len(files):
        raise RuntimeError(f'كل ملصقات {split} مفقودة؛ تحقق من تنزيل YOLO dataset وبنية labels.')

# تدقيق عدد كبير من ملفات الملصقات وتحويل الفئات إلى ID صحيح
for split, files in IMAGE_FILES.items():
    labels_dir = split_dirs[split].parent / 'labels'
    if not labels_dir.is_dir():
        continue
    bad = []
    for image_path in files[:min(5000, len(files))]:
        lp = labels_dir / image_path.relative_to(split_dirs[split]).with_suffix('.txt')
        if not lp.exists(): continue
        try:
            for line in lp.read_text(encoding='utf-8').splitlines():
                if not line.strip(): continue
                values = [float(v) for v in line.split()]
                if len(values) != 5 or int(values[0]) != values[0] or not (0 <= int(values[0]) < len(NAMES)) or not all(-0.001 <= v <= 1.001 for v in values[1:]):
                    raise ValueError('تنسيق YOLO detect غير صحيح (Class, x, y, w, h)')
        except Exception as exc: bad.append((str(lp), str(exc)))
        if len(bad) >= 5: break
    if bad: raise RuntimeError(f'ملصقات غير صالحة في {split}: {bad}')

print('✅ الفئات:', dict(enumerate(NAMES)))
print('✅ data.yaml:', TRAIN_YAML)
````

## 8 — بصمة البيانات وثبات التجربة (Data Lock)

نحسب بصمة قائمة الصور/الملصقات وأحجامها والفئات. هذا الفحص يلتقط تغيّر أسماء الملفات وتقسيمها وترتيب الفئات؛ **ليس** بصمة تشفيرية لكل بايت من جميع الصور (لتسريع التحضير).

````python
def fingerprint_dataset():
    # Stable across sessions: source identity + image paths/sizes + SHA256 for label files.
    # Image content hashing is skipped to avoid a long startup scan of all large images.
    identity = {
        'names': NAMES,
        'source': DATA_SOURCE,
        'roboflow_workspace': ROBOFLOW_WORKSPACE if DATA_SOURCE == 'roboflow' else None,
        'roboflow_project': ROBOFLOW_PROJECT if DATA_SOURCE == 'roboflow' else None,
        'roboflow_version': ROBOFLOW_VERSION if DATA_SOURCE == 'roboflow' else None,
    }
    h = hashlib.sha256(json.dumps(identity, ensure_ascii=False, sort_keys=True).encode('utf-8'))
    for split in ('train', 'val'):
        files = IMAGE_FILES[split]
        label_dir = split_dirs[split].parent / 'labels'
        for img in files:
            rel = img.relative_to(split_dirs[split]).as_posix()
            lb = label_dir / Path(rel).with_suffix('.txt')
            entry = f'{split}|{rel}|{img.stat().st_size}|'
            h.update(entry.encode('utf8'))
            if lb.exists():
                h.update(sha256_file(lb).encode('ascii'))
            else:
                h.update(b'MISSING_LABEL')
            h.update(b'\n')
    return h.hexdigest()

DATA_FINGERPRINT = fingerprint_dataset()
CURRENT_CONFIG = {
    'schema': 2,
    'run_id': RUN_ID,
    'data_source': DATA_SOURCE,
    'data_fingerprint': DATA_FINGERPRINT,
    'classes': NAMES,
    'model': MODEL_NAME,
    'epochs': EPOCHS,
    'imgsz': IMG_SIZE,
    'batch': BATCH_SIZE,
    'patience': PATIENCE,
    'workers': WORKERS,
    'seed': SEED,
    'optimizer': 'AdamW',
    'lr0': 0.001,
    'cos_lr': True,
    'amp': True,
    'deterministic': True,
    'roboflow_workspace': ROBOFLOW_WORKSPACE if DATA_SOURCE == 'roboflow' else None,
    'roboflow_project': ROBOFLOW_PROJECT if DATA_SOURCE == 'roboflow' else None,
    'roboflow_version': ROBOFLOW_VERSION if DATA_SOURCE == 'roboflow' else None,
    'ultralytics_version': ultralytics.__version__,
    'torch_version': torch.__version__,
}
IMMUTABLE_KEYS = ['data_source', 'data_fingerprint', 'classes', 'model', 'epochs', 'imgsz', 'batch', 'patience', 'seed', 'optimizer', 'lr0', 'cos_lr', 'amp', 'deterministic', 'roboflow_workspace', 'roboflow_project', 'roboflow_version']
remote_config_path = CONFIG_DIR / f'{RUN_ID}_remote_config.json'
if CLOUD_RESTORED_META is not None and CLOUD_RESTORED_META.get('data_fingerprint') != DATA_FINGERPRINT:
    raise RuntimeError('بصمة البيانات في نقطة الحفظ لا تطابق النسخة الحالية؛ أوقفنا الاستئناف لحماية النتائج.')
if remote_config_path.is_file():
    previous_config = json.loads(remote_config_path.read_text(encoding='utf8'))
    conflicts = {k: (previous_config.get(k), CURRENT_CONFIG.get(k)) for k in IMMUTABLE_KEYS if previous_config.get(k) != CURRENT_CONFIG.get(k)}
    if conflicts:
        raise RuntimeError(f'⛔ تغيرت إعدادات تجربة محفوظة على Drive: {conflicts}. أعد القيم الأصلية أو استخدم RUN_ID جديدًا.')
    if previous_config.get('ultralytics_version') != ultralytics.__version__:
        print('⚠️ تغيّر إصدار Ultralytics. إذا فشل الاستئناف ثبّت الإصدار السابق:', previous_config.get('ultralytics_version'))
    print('✅ تطابقت إعدادات التجربة مع Drive؛ يمكن الاستئناف بأمان.')
else:
    if CLOUD_RESTORED_META is not None:
        raise RuntimeError('وجدت checkpoint في Drive لكن run_config.json مفقود؛ أوقفنا التدريب لتجنب عدم تطابق البيانات.')
    print('✅ تجربة جديدة: أول حفظ لإعداداتها على Drive.')

write_json(CONFIG_DIR / f'{RUN_ID}_run_config.json', CURRENT_CONFIG)
upload_file(CONFIG_DIR / f'{RUN_ID}_run_config.json', 'meta/run_config.json')
upload_file(TRAIN_YAML, 'run/data.yaml')
print('Fingerprint:', DATA_FINGERPRINT[:20], '…')
````

## 9 — فحص جودة البيانات وتوازن الفئات

مجموعة Roboflow `ppes-kaxsi` الإصدار 8 هي **Object Detection** وبها 12 فئة. صفحة المجموعة تشير إلى أن بعض الفئات قليلة التمثيل؛ لذلك لا تعتبر عدد الصور وحده دليلًا على توازن البيانات أو الدقة المتوقعة. افحص توزيع الفئات وعيّنات الملصقات قبل اعتماد النموذج أمام اللجنة.

الفحص التالي اختياري للتحقق من الصور المتطابقة بين Train وValidation. لا يغير التقسيم تلقائيًا.

````python
# ملخص توزيع الفئات في ملفات Train (عدد الأجسام وعدد الصور التي تظهر فيها الفئة)
class_object_counts = Counter()
class_image_counts = Counter()
train_labels_dir = split_dirs['train'].parent / 'labels'
for image_path in IMAGE_FILES['train']:
    label_path = train_labels_dir / image_path.relative_to(split_dirs['train']).with_suffix('.txt')
    if not label_path.is_file():
        continue
    seen_in_image = set()
    for line in label_path.read_text(encoding='utf-8').splitlines():
        if not line.strip():
            continue
        class_id = int(float(line.split()[0]))
        class_object_counts[class_id] += 1
        seen_in_image.add(class_id)
    for class_id in seen_in_image:
        class_image_counts[class_id] += 1

class_report = {
    NAMES[class_id]: {
        'objects': int(class_object_counts[class_id]),
        'images': int(class_image_counts[class_id]),
    }
    for class_id in range(len(NAMES))
}
print('توزيع فئات Train:')
for class_name, counts in class_report.items():
    print(f"- {class_name}: {counts['objects']:,} objects / {counts['images']:,} images")
write_json(LOCAL_REPORT_DIR / 'class_distribution.json', class_report)
try:
    upload_file(LOCAL_REPORT_DIR / 'class_distribution.json', 'reports/class_distribution.json')
except Exception as exc:
    print('⚠️ تقرير توزيع الفئات محفوظ محليًا؛ تعذر رفعه:', type(exc).__name__)

if CHECK_DUPLICATE_IMAGES:
    def image_hash_index(items):
        output = {}
        for idx, p in enumerate(items, 1):
            output.setdefault(sha256_file(p), []).append(str(p))
            if idx % 5000 == 0: print('فُحصت', idx, 'صورة')
        return output
    train_hashes = image_hash_index(IMAGE_FILES['train'])
    val_hashes = image_hash_index(IMAGE_FILES['val'])
    duplicates = sorted(set(train_hashes).intersection(val_hashes))
    if duplicates:
        report = LOCAL_REPORT_DIR / 'cross_split_duplicates.json'
        write_json(report, [{'train':train_hashes[k][:2], 'val':val_hashes[k][:2]} for k in duplicates])
        print(f'⚠️ توجد {len(duplicates)} صورة مطابقة عبر التقسيمين. راجع التقرير قبل اعتماد نتائج الدقة.')
    else:
        print('✅ لا توجد صور متطابقة بايتًا بين train و val.')
else:
    print('يمكن تفعيل CHECK_DUPLICATE_IMAGES لاحقًا. لن نغير التقسيم تلقائيًا.')
````

## 10 — الحفظ السحابي عقب كل Epoch مكتمل

- بعد كل Epoch، يرفع الدفتر `last.pt` إلى اسم مؤقت ثم ينشره بعد اكتمال الرفع، مع ملف SHA-256 مستقل.
- لا تُعتبر النقطة صالحة للاستعادة إلا عند وجود ملف الوزن وملف التحقق معًا.
- يحتفظ افتراضيًا بأحدث **4** نقاط موثقة، إضافة إلى `weights/best.pt` وسجل المقاييس.
- عند فشل رفع checkpoint، يحاول الدفتر مجددًا مرتين. إذا استمر الفشل، **يوقف التدريب** بدل إكمال حقب لا يمكن ضمان حفظها.
- لا يُحفظ كل Batch؛ الاستعادة تكون من آخر Epoch مكتمل وصل إلى Drive.

````python
def prune_old_cloud_snapshots():
    try:
        paths = cloud_paths()
        records = remote_checkpoint_records(paths)
        for ep, meta_rel, weight_rel in records[KEEP_CLOUD_SNAPSHOTS:]:
            run_rclone(['deletefile', f'{CLOUD_RUN}/{meta_rel}'])
            run_rclone(['deletefile', f'{CLOUD_RUN}/{weight_rel}'])
    except Exception as ex:
        print('⚠️ تعذر تنظيف النسخ القديمة؛ نحتفظ بها للأمان:', type(ex).__name__)

LAST_UPLOADED_BEST_SHA = None

def upload_best_if_changed():
    global LAST_UPLOADED_BEST_SHA
    path = WEIGHTS_DIR / 'best.pt'
    if not path.is_file():
        return
    actual = sha256_file(path)
    if actual == LAST_UPLOADED_BEST_SHA:
        return
    upload_atomic(path, 'weights/best.pt')
    LAST_UPLOADED_BEST_SHA = actual

def sync_training_epoch(trainer):
    # on_model_save runs after the epoch checkpoint has been written locally.
    epoch = int(trainer.epoch) + 1
    ckpt = Path(trainer.last)
    if not ckpt.is_file():
        raise RuntimeError(f'لا يوجد last.pt بعد Epoch {epoch}؛ أوقفنا التدريب لحماية التقدم.')

    remote_weight = f'checkpoints/epoch_{epoch:04d}.pt'
    stopped = bool(getattr(trainer, 'stop', False))
    metadata = {
        'schema': 2,
        'run_id': RUN_ID,
        'epoch': epoch,
        'file': remote_weight,
        'bytes': ckpt.stat().st_size,
        'sha256': sha256_file(ckpt),
        'data_fingerprint': DATA_FINGERPRINT,
        'time_utc': datetime.datetime.now(datetime.timezone.utc).isoformat(),
        'fitness': float(trainer.fitness) if getattr(trainer, 'fitness', None) is not None else None,
        'training_complete': stopped,
        'stop_reason': 'final_or_early_stopping' if stopped else None,
    }
    manifest = LOCAL_REPORT_DIR / f'epoch_{epoch:04d}.json'
    write_json(manifest, metadata)

    # A point is valid only when both the weight and its SHA256 manifest are published.
    last_error = None
    synced = False
    for attempt in range(1, SYNC_RETRIES + 2):
        try:
            upload_atomic(ckpt, remote_weight)
            upload_atomic(manifest, f'checkpoints/epoch_{epoch:04d}.json')
            synced = True
            break
        except Exception as exc:
            last_error = exc
            print(f'⚠️ تعذرت مزامنة Epoch {epoch} (محاولة {attempt}/{SYNC_RETRIES + 1}): {type(exc).__name__}')
            if attempt <= SYNC_RETRIES:
                time.sleep(SYNC_RETRY_DELAY_SECONDS * attempt)

    if not synced:
        log_event('sync_failed_stop', epoch=epoch, error_type=type(last_error).__name__ if last_error else 'Unknown')
        raise RuntimeError(
            f'توقف التدريب بعد Epoch {epoch}: لم نستطع تأكيد حفظ checkpoint على Google Drive. '
            'لا تغلق الجلسة الآن إن أمكن؛ أعد الاتصال ثم شغّل خلايا الاستعادة. '
            'لن نواصل التدريب على تقدم غير محمي.'
        ) from last_error

    log_event('cloud_synced', epoch=epoch, sha256=metadata['sha256'], bytes=metadata['bytes'])
    print(f'☁️✅ Epoch {epoch} محفوظ ومتحقق على Drive (SHA256: {metadata["sha256"][:12]}…)')

    # latest.json and best.pt are useful, but the verified per-epoch pair is the recovery source.
    try:
        upload_atomic(manifest, 'meta/latest.json')
    except Exception as ex:
        print('⚠️ تعذر تحديث latest.json؛ الاستعادة ستبحث عن أحدث checkpoint موثق:', type(ex).__name__)
    try:
        upload_best_if_changed()
        for filename in ['args.yaml', 'results.csv']:
            p = RUN_DIR / filename
            if p.is_file():
                upload_atomic(p, f'run/{filename}')
    except Exception as optional_exc:
        print('⚠️ checkpoint محفوظ، لكن تعذرت مزامنة بعض التقارير/أفضل وزن:', type(optional_exc).__name__)

    if epoch % 4 == 0:
        prune_old_cloud_snapshots()

def final_sync_best():
    try:
        upload_best_if_changed()
        for filename in ['args.yaml', 'results.csv']:
            p = RUN_DIR / filename
            if p.is_file():
                upload_atomic(p, f'run/{filename}')
        log_path = LOCAL_REPORT_DIR / 'events.jsonl'
        if log_path.is_file():
            upload_atomic(log_path, 'meta/events.jsonl')
        prune_old_cloud_snapshots()
        print('✅ المزامنة الختامية اكتملت.')
    except Exception as ex:
        print('⚠️ تعذر رفع جزء من المخرجات الختامية:', type(ex).__name__)

print('✅ Callback الحفظ الدوري جاهز: حفظ كل Epoch، محاولات إعادة، وإيقاف آمن عند فشل المزامنة.')
````

## 11 — تشغيل التدريب أو استئنافه تلقائيًا

إذا وُجدت نقطة `last.pt` موثقة على Drive، يستعيد الدفتر الأوزان وحالة optimizer ورقم Epoch ثم يستأنف نفس التجربة. إذا اكتشف أن التدريب اكتمل أو توقف مبكرًا، ينتقل إلى التقييم ولا يعيد تشغيله. إذا كانت نقطة الحفظ تالفة، يجرب الأقدم؛ وإذا لم توجد نقطة سليمة، يتوقف دون بدء تدريب جديد تلقائيًا.

`resume=True` مخصص لاستكمال تدريب انقطع ولم تتغير بياناته أو إعداداته. عند تغيير البيانات أو الفئات أو إعدادات التدريب، أنشئ `RUN_ID` جديدًا.

````python
if not torch.cuda.is_available():
    raise RuntimeError('⛔ GPU غير مفعّل. افتح Kaggle Settings واختر GPU قبل تنفيذ التدريب.')

last_path = WEIGHTS_DIR / 'last.pt'
# تحقق من وجود checkpoint في الذاكرة بعد الاستعادة
IS_RESUME = (CLOUD_RESTORED_META is not None) or last_path.is_file()
if IS_RESUME and not last_path.exists():
    raise RuntimeError('استعيدت بيانات نقطة سحابية لكن last.pt غير موجود')

if IS_RESUME:
    # فحص مبكر لحالة checkpoint: Optimizer قد يُزال عند إنهاء التدريب بالكامل
    from ultralytics.utils import torch_utils
    try:
        ckpt_data = torch.load(last_path, map_location='cpu', weights_only=False)
        saved_epoch = int(ckpt_data.get('epoch', -1)) + 1
        has_optimizer = ckpt_data.get('optimizer') is not None
        del ckpt_data
    except Exception as exc:
        raise RuntimeError('Checkpoint تم التحقق من نقله، لكنه غير قابل للقراءة بواسطة PyTorch.') from exc
    if CLOUD_RESTORED_META is not None and CLOUD_RESTORED_META.get('training_complete'):
        print(f'✅ سجل Drive يوضح أن التدريب اكتمل/توقف مبكرًا عند Epoch {saved_epoch}. لن نعيد تشغيله.')
        SHOULD_TRAIN = False
    elif saved_epoch >= EPOCHS:
        print(f'✅ التدريب بلغ {saved_epoch}/{EPOCHS} Epoch. لا حاجة لإعادة التدريب.')
        SHOULD_TRAIN = False
    elif not has_optimizer:
        raise RuntimeError('checkpoint لا يحتوي optimizer قبل نهاية خطة التدريب. لا تبدأ من الصفر؛ راجع best.pt أو أنشئ RUN_ID جديدًا.')
    else:
        print(f'🔄 استكمال بعد Epoch {saved_epoch}/{EPOCHS} باستخدام last.pt')
        SHOULD_TRAIN = True
else:
    print('🆕 تجربة جديدة؛ سيبدأ التدريب ويُحفَظ كل Epoch على Google Drive.')
    SHOULD_TRAIN = True

if SHOULD_TRAIN:
    model = YOLO(str(last_path) if IS_RESUME else MODEL_NAME)
    model.add_callback('on_model_save', sync_training_epoch)
    if IS_RESUME:
        result = model.train(resume=True, data=str(TRAIN_YAML), device=0)
    else:
        result = model.train(
            data=str(TRAIN_YAML),
            epochs=EPOCHS,
            imgsz=IMG_SIZE,
            batch=BATCH_SIZE,
            device=0,
            workers=WORKERS,
            patience=PATIENCE,
            project=str(RUNS_DIR),
            name=RUN_ID,
            exist_ok=True,
            optimizer='AdamW',
            lr0=0.001,
            cos_lr=True,
            amp=True,
            seed=SEED,
            deterministic=True,
            cache=False,
            save=True,
            save_period=5,
            plots=True,
            verbose=True,
        )
    final_sync_best()
else:
    print('لا يوجد تدريب مطلوب؛ الانتقال إلى التقييم.')
````

## 12 — ملخص النتائج + التحقق على Validation

يقارن التقرير قيمة `mAP50-95` على تقسيم التحقق. لا تساوي هذه القيمة دقة النظام بالكامل في الواقع؛ يلزم أيضًا تقييم مقاطع من بيئة المشروع، الفئات المفقودة، ومعدل الإنذارات الخاطئة.

````python
BEST_WEIGHTS = WEIGHTS_DIR / 'best.pt'
if not BEST_WEIGHTS.is_file() and (WEIGHTS_DIR / 'last.pt').is_file():
    print('⚠️ لم أجد best.pt؛ الاستعاضة مؤقتًا بـ last.pt للتقييم.')
    BEST_WEIGHTS = WEIGHTS_DIR / 'last.pt'

if BEST_WEIGHTS.is_file():
    print('وزن التقييم:', BEST_WEIGHTS, '| MB:', round(BEST_WEIGHTS.stat().st_size/1024**2, 2))
    if RUN_VALIDATION_AFTER_TRAINING:
        evaluation_model = YOLO(str(BEST_WEIGHTS))
        metrics = evaluation_model.val(data=str(TRAIN_YAML), imgsz=IMG_SIZE, batch=1, device=0 if torch.cuda.is_available() else 'cpu', project=str(ROOT/'validation'), name=RUN_ID, exist_ok=True)
        report = {
            'run_id': RUN_ID,
            'weights': BEST_WEIGHTS.name,
            'data_fingerprint': DATA_FINGERPRINT,
            'mAP50': float(metrics.box.map50),
            'mAP50_95': float(metrics.box.map),
            'precision': float(metrics.box.mp),
            'recall': float(metrics.box.mr),
            'classes': NAMES,
            'time_utc': datetime.datetime.now(datetime.timezone.utc).isoformat(),
        }
        write_json(LOCAL_REPORT_DIR/'validation_summary.json', report)
        print(json.dumps(report, indent=2, ensure_ascii=False))
        try:
            upload_file(LOCAL_REPORT_DIR/'validation_summary.json', 'reports/validation_summary.json')
        except Exception as exc:
            print('⚠️ تقرير التقييم محفوظ محليًا؛ تعذر رفعه:', type(exc).__name__)
else:
    print('لا توجد أوزان للتقييم بعد؛ نفّذ خلية التدريب أولًا.')
````
